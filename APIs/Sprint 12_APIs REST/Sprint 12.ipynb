##### Ejercicio 1
Consulta la API pública JSONPlaceholder utilizando el método GET para obtener:

- Lista de publicaciones (/posts)
- Lista de usuarios (/users)
- Lista de tareas (/todos)

##### Ejercicio 2
Muestra por pantalla:

- La cantidad total de cada recurso.
- El código de estado de cada petición.

```python
#Para obtener lista de publicaciones (/posts):
lista_publis=requests.get("https://jsonplaceholder.typicode.com/posts")
display(f"Hay {len(lista_publis.json())} publicaciones y su codigo de petición es: {lista_publis.status_code}")


#Para obtener lista de usuarios (/users)
lista_usuarios=requests.get("https://jsonplaceholder.typicode.com/users")
display(f"Hay {len(lista_usuarios.json())} publicaciones y su codigo de petición es: {lista_usuarios.status_code}")

#Para obtener lista de tareas (/todos)
lista_todos=requests.get("https://jsonplaceholder.typicode.com/todos")
display(f"Hay {len(lista_todos.json())} publicaciones y su codigo de petición es: {lista_todos.status_code}")
```

##### Ejercicio 3:
Realiza una petición a una publicación inexistente para obtener un error 404 y muestra el código de estado recibido.

##### Ejercicio 4:
Realiza una petición POST para crear una nueva publicación ficticia. 
Incluye el título, el cuerpo del mensaje y un userId.

- Muestra la respuesta JSON.
- Muestra el código de estado.

```python
publi_ficti=requests.post("https://jsonplaceholder.typicode.com/posts", data={"titulo": "sprint 12",
                                                                  "cuerpo":"aprendiendo APIs",
                                                                  "userID": 101})

display(f"Respuesta JSON del servidor: {publi_ficti.json()}, Código de estado: {publi_ficti.status_code}")
```

##### Ejercicio 5:
Realiza una petición PATCH para modificar parcialmente una publicación existente.

- Muestra la respuesta JSON.
- Muestra el código de estado.


```python
publi_modif=requests.patch("https://jsonplaceholder.typicode.com/posts/101", data={"titulo":"sprint 12, modificado"})

display(f"Respuesta del servidor: {publi_modif.json()}, Código de estado: {publi_modif.status_code}")
```

##### Ejercicio 6:
Haz una petición DELETE sobre una publicación.

- Muestra la respuesta JSON.
- Muestra el código de estado.


```python
eliminar_publi=requests.delete("https://jsonplaceholder.typicode.com/posts/36")

display(f"Respuesta del servidor: {eliminar_publi.json()}, Código de estado: {eliminar_publi.status_code}")
```
