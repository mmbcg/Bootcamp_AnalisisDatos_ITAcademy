1. Calculadora de indice de masa corporal:

- Escribir una función que calcule el IMC ingresado por el usuario/a, es decir,
    quien lo ejecute deberá ingresar sus datos.

- La función debe clasificar el resultado en sus respectivas categorías.

Consejo: Validar los datos primero, que envie un mensaje de advetencia 
        si los datos introducidos no estan en el formato adecuado o no toma valores razonables.

        --> que sean numeros, que no sean negativos y que tengan sentido para pesos y alturas

Notas:

IMC= peso en kg / altura al cuadrado en metros

de esto se obtienen las siguietes categorias:
- Bajo peso: menos de 18.5
- Peso normal: 18.5-24.9
- Sobrepeso: 25-29.9
- Obesidad: más de 30


```python
def calculadora_imc():
    try:
        #solicitar datos al usuario:
        peso=float(input("Introducir peso en kg:"))
        altura=float(input("Introducir altura en metros:"))

        #Validación de datos ingresados
        if peso <=0 or altura <=0:
            print("Error!: peso y altura deben ser valores positivos")
            return
        if peso <20 or peso > 300:
            print("Error!: El peso introducido no es razonable")
            return
        if altura <0.5 or altura > 2.5:
            print("Error!: La altura introducida no es razonable")
            return
        #Calculo del IMC
        imc=peso/(altura**2)

        #clasificacion del IMC:
        if imc <18.5:
            categoria="Bajo peso"
        elif imc<25:
            categoria="Peso normal"
        elif imc<30:
            categoria="Soobrepeso"
        else:
            categoria="Obesidad"
        
        #resultados:
        print(f"Tu IMC es de {imc:.2f}")
        print(f"Clasificación:{categoria}")

    except ValueError:
        print("Error!: Por favor, introducir valores númericos validos.")


calculadora_imc()
```

2. Conversor de temperaturas 
- Seleccionar al menos dos conversores, es decir,
    al introducir una temperatura devuelva como minimo 2 conversiones 
- Guardar las conversiones que se van introduciendo
    * Friendly reminder: el print no se puede guardar nunca!!!!!
- Validar los datos previamentee
- almacenar todas las posible conversiones en un solo objeto --> para no usar muchos if else

Vale pues:
- °C-->°F: (C * 9/5) + 32
- °C-->K: C + 273.15
- °F-->°C: (F - 32) * 5/9
- K-->°C: K - 273.15
** todas las conversiones pasan por °C

```python
def conversor_temperatura():
    try:
        #pedir datos:
        temperatura=float(input("Introducir temperatura:"))
        unidad=input("Indicar la unidad de origen(C, F, K:)").upper()

        #validacion de unidad:
        if unidad not in ["C", "F", "K"]:
            print("Error!: unidad no válida. Usar C, F o K.")
            return
        
        #conversion a Ceelsius:
        if unidad =="C":
            temp_c=temperatura
        elif unidad == "F":
            temp_c=(temperatura-32)*5/9
        else: #K
            temp_c=temperatura-273.15
        
        #creamos el diccionario para guardar las conversiones
        conversiones={
            "Celsius(°C)": round(temp_c,2),
            "Fahrenheit(°F)": round((temp_c*9/5)+32,2),
            "Kelvin (K)":round(temp_c+273.15,2)
        }

        return conversiones
    except ValueError:
        print("Error!: Introducir un valor númerico válido.")

conversor_temperatura()
```

3. Contador de palabras de un texto:

Escribe una función que dado un texto, muestre las veces que aparece cada palabra.
Intenta que se gestionen todas las posibles casuísticas que hagan que el programa no funcione correctamente.

(EXTRA): ¿Cuál es la longitud media de las palabras del texto que has escrito? “¿Hola cómo va?” debería devolver (4+3+2) / 3 = 3





```python
def contador_palabras(texto):
    if not isinstance(texto,str):
        print("Error!: el texto debe ser tipo cadena(str)")
        return
    if texto.strip()=="":
        print("Error!: el texto está vacio")
        return
    texto=texto.lower()
    signos="¿?¡!.,;:"
    for signo in signos:
        texto=texto.replace(signo,"")
    palabras=texto.split()

    conteo={}
    total_letras=0

    for palabra in palabras:
        conteo[palabra]=conteo.get(palabra,0)+1
        total_letras+=len(palabra)

    longitud_media=total_letras/len(palabras)

    return conteo, longitud_media

conteo, media=contador_palabras("Hola, buenos dias princesa!")

print(conteo)
print(media)
```

4. Diccionario inverso (con posibilidad de duplicados)

Resulta que el cliente tiene una encuesta muy antigua que se almacena en un diccionario y los resultados los necesita a la inversa, es decir,
intercambiando las claves y valores. Los valores y claves en el diccionario original son únicos; si éste no es el caso, 
la función debería imprimir un mensaje de advertencia, junto con una lista con los valores asociados a la clave repetida.

- Fijarse en los ejemplos para ver el output esperado


```python
def diccionario_inverso(diccionario):
    if not isinstance(diccionario,dict):
        print("Error: el argumento debe ser un diccionario")
        return
    
    invertido={}

    for clave, valor in diccionario.items():
        #si el valor no existe todavia:
        if valor not in invertido:
            invertido[valor]=clave

        #si el valor ya exite en el diccionario:
        else:
            #se convierte
            if not isinstance(invertido[valor],list):
                invertido[valor]=[invertido[valor]]

            invertido[valor].append(clave)

    return invertido


resultado=diccionario_inverso({'x': 'apple', 'y': 'banana', 'z': 'banana'})
print(resultado)
```

1. Contador y ordenador de palabras de un texto.

Leer archivos TXT y calcular la frecuencia de cada palabra ordenada dentro de las entradas habituales del diccionario 
según la letra con la que comienzan, es decir, 

las claves deben ir de la A a la Z y dentro de la A debemos ir de la A a Z.



```python
#me da toc y me gusta que se veo vertical el resultado:
from pprint import pprint


#crear la funcion
def contador_palabras_archivo(ruta_archivo):

    #Para leer el doc .txt:
    with open(ruta_archivo, encoding="utf-8") as f:
        texto = f.read()

    #limpiar el texto
    texto = texto.lower()
    signos = "¿?¡!.,;:()[]\"'"
    for signo in signos:
        texto = texto.replace(signo, "")

    palabras = texto.split()

    #Para contar la frecuencia de cada palabra:
    frecuencia = {}
    for palabra in palabras:
        frecuencia[palabra] = frecuencia.get(palabra, 0) + 1

    #por letra principal de cada palabra:
    resultado = {}
    for palabra, veces in frecuencia.items():
        letra = palabra[0]
        if letra not in resultado:
            resultado[letra] = {}
        resultado[letra][palabra] = veces


    #Orden alfabetico:
    resultado_ordenado = {}
    for letra in sorted(resultado.keys()):
        resultado_ordenado[letra] = dict(sorted(resultado[letra].items()))

    return resultado_ordenado

#pa ve:
diccionario = contador_palabras_archivo("tu me quieres blanca.txt")
pprint(diccionario)

```

```python
def conversion(lista):
    convertibles = []
    no_convertibles = []

    def procesar(elemento):
        # Si es lista o tupla → recorrer
        if isinstance(elemento, (list, tuple)):
            for item in elemento:
                procesar(item)
        else:
            try:
                convertibles.append(float(elemento))
            except (ValueError, TypeError):
                no_convertibles.append(elemento)

    procesar(lista)
    return convertibles, no_convertibles

datos = [
    '1.3', 'one', '1e10', 'seven', '3-1/2',
    ('2', 1, 1.4, 'not-a-number'),
    [1, 2, '3', '3.4']
]

resultado = conversion(datos)
print(resultado)
```
